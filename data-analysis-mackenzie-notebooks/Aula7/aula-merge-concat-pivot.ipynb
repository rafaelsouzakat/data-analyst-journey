Tratamento de dados: junção, combinação e reformatação

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def __repr__(self):
        return f"{self.nome} - R$ {self.preco:.2f}"


class Subcategoria:
    def __init__(self, nome):
        self.nome = nome
        self.produtos = []

    def adicionar_produto(self, produto):
        self.produtos.append(produto)

    def exibir_produtos(self):
        print(f"Produtos em {self.nome}:")
        for produto in self.produtos:
            print(f"  - {produto}")


class Categoria:
    def __init__(self, nome):
        self.nome = nome
        self.subcategorias = []

    def adicionar_subcategoria(self, subcategoria):
        self.subcategorias.append(subcategoria)

    def exibir_subcategorias(self):
        print(f"Categoria: {self.nome}")
        for subcategoria in self.subcategorias:
            subcategoria.exibir_produtos()


# Criando categorias e subcategorias
eletronicos = Categoria("Eletrônicos")
celulares = Subcategoria("Celulares")
celulares.adicionar_produto(Produto("iPhone 14", 6999.99))
celulares.adicionar_produto(Produto("Samsung Galaxy S22", 4999.99))

notebooks = Subcategoria("Notebooks")
notebooks.adicionar_produto(Produto("Dell XPS 13", 5999.99))
notebooks.adicionar_produto(Produto("MacBook Pro", 12999.99))

eletronicos.adicionar_subcategoria(celulares)
eletronicos.adicionar_subcategoria(notebooks)

# Exibindo a estrutura
eletronicos.exibir_subcategorias()
```

```python
class Product:
    def __init__(self, name, category):
        self.name = name
        self.category = category

    def __repr__(self):
        return f"Product(name='{self.name}', category='{self.category}')"

# The ProductIndex class should be defined outside the Product class
class ProductIndex:
    def __init__(self):
        self.products = []
        self.index = {}

    def add_product(self, product):
        self.products.append(product)
        # Indexação parcial: apenas indexa produtos da categoria "eletrônicos"
        if product.category == "eletrônicos":
            self.index[product.name] = product

    def search(self, name):
        return self.index.get(name, None)

# Criando o índice de produtos
product_index = ProductIndex()

# Adicionando produtos
product_index.add_product(Product("iPhone 14", "eletrônicos"))
product_index.add_product(Product("Camiseta", "vestuário"))
product_index.add_product(Product("Samsung Galaxy S22", "eletrônicos"))
product_index.add_product(Product("Calça Jeans", "vestuário"))

# Buscando produtos indexados
print(product_index.search("iPhone 14"))  # Saída: <__main__.Product object at ...>
```

MultiIndex: estrutura de índice que permite a criação de índices hierárquicos em um DataFrame ou Series. Isso significa que você pode ter múltiplos níveis de índices, o que é útil para organizar e acessar dados de forma mais complexa e flexível.
Quando Usar MultiIndex
Quando seus dados têm uma estrutura hierárquica natural (por exemplo, dados de vendas por produto e por região).
Quando você precisa realizar operações de agregação ou resumo em múltiplos níveis de dados.
Um MultiIndex pode aumentar a complexidade do seu código, por isso, use-o quando realmente necessário.
A manipulação de MultiIndex pode ser um pouco mais desafiadora, mas oferece muita flexibilidade e poder para a análise de dados.

O método swaplevel()

```python
import pandas as pd

# Criar um DataFrame com MultiIndex
arrays = [
    ['A', 'A', 'B', 'B'],
    [1, 2, 1, 2]
]
index = pd.MultiIndex.from_arrays(arrays, names=('letter', 'number'))
df = pd.DataFrame({'value': [10, 20, 30, 40]}, index=index)

print("DataFrame Original:")
print(df)

# Trocar os níveis do índice
df_swapped = df.swaplevel('letter', 'number')

print("\nDataFrame com Níveis Trocados:")
print(df_swapped)
```

sort_index() no pandas é utilizado para ordenar um DataFrame ou uma Series com base nos índices. Isso pode ser útil quando você deseja visualizar os dados de uma forma mais organizada ou preparar os dados para operações subsequentes.

DataFrame.sort_index(axis=0, level=None, ascending=True, inplace=False, kind='quicksort', na_position='last', sort_remaining=True)
axis: O eixo ao longo do qual ordenar (0 para linhas, 1 para colunas). O padrão é 0.
level: Se o índice for um MultiIndex, você pode especificar qual nível deve ser ordenado.
ascending: Um booleano que indica se a ordenação deve ser crescente (True) ou decrescente (False). O padrão é True.
inplace: Se True, realiza a operação no próprio DataFrame/Series, sem criar uma cópia. O padrão é False.
na_position: Pode ser 'last' (padrão) ou 'first', indicando se os valores NaN devem aparecer no início ou no final.
kind: O algoritmo de ordenação a ser usado (por exemplo, 'quicksort' - velocidade, 'mergesort' - ordenação, 'heapsort' - desempenho).

```python
import pandas as pd

# Criar um DataFrame
data = {
    'A': [3, 1, 2],
    'B': ['C', 'A', 'B']
}
df = pd.DataFrame(data, index=['row2', 'row1', 'row3'])

print("DataFrame Original:")
print(df)

# Ordenar pelo índice
df_sorted = df.sort_index()

print("\nDataFrame Ordenado pelo Índice:")
print(df_sorted)
```

```python
import pandas as pd

# Criar um DataFrame
data = {
    'A': [3, 1, 2],
    'B': ['C', 'A', 'B']
}
df = pd.DataFrame(data, index=['row2', 'row1', 'row3'])

# Ordenar pelo índice usando diferentes métodos
sorted_quick = df.sort_index(kind='quicksort')
sorted_merge = df.sort_index(kind='mergesort')
sorted_heap = df.sort_index(kind='heapsort')

print("\nDataFrame Original:")
print(df)

print("Ordenado com Quicksort:")
print(sorted_quick)

print("\nOrdenado com Mergesort:")
print(sorted_merge)

print("\nOrdenado com Heapsort:")
print(sorted_heap)
```

groupby() no pandas é uma ferramenta poderosa para realizar operações de agregação e análise em conjuntos de dados. Ele permite que você agrupe dados com base em uma ou mais colunas e, em seguida, aplique funções de agregação, transformação ou filtragem.

```python
import pandas as pd

# Criar um DataFrame de exemplo
data = {
    'Categoria': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Valores': [10, 20, 30, 40, 50, 60]
}
df = pd.DataFrame(data)

print("DataFrame Original:")
print(df)

# Agrupar por 'Categoria' e calcular a soma dos 'Valores'
grouped = df.groupby('Categoria').sum()

print("\nDataFrame Agrupado pela Soma dos Valores:")
print(grouped)

```

set_index() no pandas é utilizado para definir uma ou mais colunas de um DataFrame como o índice. Alterar o índice pode facilitar a manipulação, acesso e análise dos dados, especialmente em operações de agrupamento e filtragem.

```python
import pandas as pd

# Criar um DataFrame de exemplo
data = {
    'Nome': ['Alice', 'Bob', 'Charlie', 'David'],
    'Idade': [25, 30, 35, 40],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Curitiba']
}
df = pd.DataFrame(data)

print("DataFrame Original:")
print(df)

# Definir a coluna 'Nome' como índice
df_indexed = df.set_index('Nome')

print("\nDataFrame com 'Nome' como Índice:")
print(df_indexed)
```

```python
# Redefinir o índice sem manter a coluna
df_reset_drop = df_indexed.reset_index(drop=True)

print("\nDataFrame após Resetar o Índice (sem manter a coluna):")
print(df_reset_drop)
```

A função merge() é usada para combinar dois DataFrames com base em uma ou mais chaves comuns (similar a um JOIN em SQL).

```python
import pandas as pd

# Criar DataFrames de exemplo
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte']
})

print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)

# Realizar o merge
merged_df = pd.merge(df1, df2, on='ID', how='inner')

print("\nDataFrame após Merge:")
print(merged_df)
```

A função concat() é usada para concatenar (unir) DataFrames ao longo de um eixo (horizontal ou vertical).

```python
# Criar DataFrames de exemplo
df3 = pd.DataFrame({
    'ID': [1, 2],
    'Nome': ['Alice', 'Bob']
})

df4 = pd.DataFrame({
    'ID': [3, 4],
    'Nome': ['Charlie', 'David']
})

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Concatenar DataFrames verticalmente
concat_df = pd.concat([df3, df4], axis=0)

print("\nDataFrame após Concatenar:")
print(concat_df)
```

O método combine_first() é usado para combinar dois DataFrames, preenchendo os valores ausentes (NaN) no primeiro DataFrame com os valores do segundo DataFrame.

```python
# Criar DataFrames com valores ausentes
df5 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': ['Alice', None, 'Charlie']
})

df6 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': [None, 'Bob', 'David']
})

print("\nDataFrame 5:")
print(df5)

print("\nDataFrame 6:")
print(df6)

# Combinar DataFrames
combined_df = df5.combine_first(df6)

print("\nDataFrame após combine_first:")
print(combined_df)
```

Inner Join (how='inner'):

Retorna apenas as linhas que têm correspondências em ambos os DataFrames.
É o padrão para merge().

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Nome': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Cidade': ['Rio de Janeiro', 'São Paulo', 'Curitiba']
})

inner_join = pd.merge(df1, df2, on='ID', how='inner')

print("Inner Join:")
print(inner_join)
```

Outer Join (how='outer'):

Retorna todas as linhas de ambos os DataFrames, preenchendo com NaN onde não há correspondência.

```python
outer_join = pd.merge(df1, df2, on='ID', how='outer')

print("\nOuter Join:")
print(outer_join)
```

Left Join (how='left'):

Retorna todas as linhas do DataFrame da esquerda e as linhas correspondentes do DataFrame da direita. Se não houver correspondência, retorna NaN para as colunas do DataFrame da direita.

```python
left_join = pd.merge(df1, df2, on='ID', how='left')

print("\nLeft Join:")
print(left_join)
```

Right Join (how='right'):

Retorna todas as linhas do DataFrame da direita e as linhas correspondentes do DataFrame da esquerda. Se não houver correspondência, retorna NaN para as colunas do DataFrame da esquerda.

```python
right_join = pd.merge(df1, df2, on='ID', how='right')

print("\nRight Join:")
print(right_join)
```

REFORMATAÇÃO OU PIVOTAMENTO

pivot()
O método pivot() é usado para reorganizar os dados, transformando colunas em índices e colunas de valores.

```python
import pandas as pd

# Criar um DataFrame de exemplo
data = {
    'Ano': [2020, 2020, 2021, 2021],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'São Paulo', 'Rio de Janeiro'],
    'População': [12.33, 6.75, 12.40, 6.80]
}
df = pd.DataFrame(data)

print("DataFrame Original:")
print(df)

# Pivotar o DataFrame
pivot_df = df.pivot(index='Ano', columns='Cidade', values='População')

print("\nDataFrame Pivotado:")
print(pivot_df)
```

pivot_table()
O método pivot_table() é semelhante ao pivot(), mas permite realizar operações de agregação (como soma, média) em caso de duplicatas.

```python
# Criar um DataFrame com duplicatas
data2 = {
    'Ano': [2020, 2020, 2021, 2021, 2021],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'São Paulo', 'Rio de Janeiro', 'São Paulo'],
    'População': [12.33, 6.75, 12.40, 6.80, 12.50]
}
df2 = pd.DataFrame(data2)

print("\nDataFrame com Duplicatas:")
print(df2)

# Criar uma tabela dinâmica
pivot_table_df = df2.pivot_table(index='Ano', columns='Cidade', values='População', aggfunc='mean')

print("\nTabela Dinâmica:")
print(pivot_table_df)
```

melt()
O método melt() é usado para transformar um DataFrame de formato largo para formato longo. Isso é útil para análises onde as variáveis são melhor tratadas em uma estrutura mais compacta.

```python
# Criar um DataFrame largo
data3 = {
    'Ano': [2020, 2021],
    'São Paulo': [12.33, 12.40],
    'Rio de Janeiro': [6.75, 6.80]
}
df3 = pd.DataFrame(data3)

print("\nDataFrame Largo:")
print(df3)

# Transformar para formato longo
melted_df = df3.melt(id_vars='Ano', var_name='Cidade', value_name='População')

print("\nDataFrame Transformado para Formato Longo:")
print(melted_df)
```

```python
import pandas as pd

# DataFrame de informações dos alunos
df_alunos = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Nome': ['Alice', 'Bob', 'Charlie', 'David'],
    'Idade': [20, 21, 19, 22]
})

# DataFrame de notas
df_notas = pd.DataFrame({
    'ID': [1, 2, 3, 1, 2, 3],
    'Disciplina': ['Matemática', 'Matemática', 'Matemática', 'Física', 'Física', 'Física'],
    'Nota': [85, 90, 78, 88, 95, 80]
})

print("DataFrame de Alunos:")
print(df_alunos)
print("\nDataFrame de Notas:")
print(df_notas)

# Junção dos DataFrames
df_merged = pd.merge(df_alunos, df_notas, on='ID', how='inner')
print("\nDataFrame após Junção:")
print(df_merged)

# Calcular a média das notas por aluno
df_media = df_merged.groupby(['ID', 'Nome']).agg({'Nota': 'mean'}).reset_index()
print("\nDataFrame com Média das Notas:")
print(df_media)

# Reformatar os dados para um formato largo
df_pivot = df_merged.pivot_table(index=['ID', 'Nome', 'Idade'],
                                  columns='Disciplina',
                                  values='Nota',
                                  aggfunc='mean').reset_index()
print("\nDataFrame Reformulado:")
print(df_pivot)
```

```python
def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort(left) + middle + quicksort(right)

# Exemplo de uso
array = [10, 5, 2, 3, 8, 6, 1]
sorted_array = quicksort(array)
print("QuickSort:", sorted_array)
```

```python
def mergesort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = mergesort(arr[:mid])
    right = mergesort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    sorted_arr = []
    while left and right:
        if left[0] < right[0]:
            sorted_arr.append(left.pop(0))
        else:
            sorted_arr.append(right.pop(0))
    sorted_arr.extend(left or right)
    return sorted_arr

# Exemplo de uso
array = [10, 5, 2, 3, 8, 6, 1]
sorted_array = mergesort(array)
print("MergeSort:", sorted_array)
```

```python
def heapsort(arr):
    n = len(arr)

    # Constrói um heap máximo
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Um a um, extrai elementos do heap
    for i in range(n-1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # Troca
        heapify(arr, i, 0)

def heapify(arr, n, i):
    largest = i  # Inicializa o maior como raiz
    left = 2 * i + 1  # filho à esquerda
    right = 2 * i + 2  # filho à direita

    # Se o filho à esquerda for maior que a raiz
    if left < n and arr[left] > arr[largest]:
        largest = left

    # Se o filho à direita for maior que o maior até agora
    if right < n and arr[right] > arr[largest]:
        largest = right

    # Se o maior não for a raiz
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # Troca
        heapify(arr, n, largest)

# Exemplo de uso
array = [10, 5, 2, 3, 8, 6, 1]
heapsort(array)
print("HeapSort:", array)
```

QuickSort: Divide o array em torno de um pivô e ordena recursivamente as partes.
MergeSort: Divide o array em metades, ordena cada metade recursivamente e as mescla.
HeapSort: Constrói um heap máximo e extrai elementos em ordem crescente.
