```python
# encontrar colunas com dados do tipo String ou object
string_columns = companies_df.select_dtypes(include='object').columns

# aplicar .str.lower() para todas colunas com String
companies_df[string_columns] = companies_df[string_columns].apply(lambda x: x.str.lower() if x.dtype == 'object' else x, axis=0)

# mostrar dataframe para ver mudanças
companies_df.head()
```

```python
# Remove caracteres especiais de colunas com Strings
companies_df[string_columns] = companies_df[string_columns].apply(lambda x: x.str.replace(r'[^a-z0-9\s]', '', regex=True))
```

```python
def parse_marketcap(marketcap_str):
    if pd.isna(marketcap_str) or not isinstance(marketcap_str, str):
        return np.nan
    marketcap_str = marketcap_str.replace('$', '').strip().lower()
    # cuida de casos em que a unidade está grudada com o número
    try:
        value = float(marketcap_str[:-1])
        unit = marketcap_str[-1]
    except ValueError:
        return np.nan # se falhar, retorna NaN

    if unit == 't':
        return value * 1000 # converte trilhões para bilhões
    elif unit == 'b':
        return value # valor já é em bilhões
    elif unit == 'm':
        return value / 1000 # Converte milhões para bilhões
    else:
        return np.nan # cuida de casos em que a unidade não é t, b ou m
```

```python
companies_df['marketcap_billion'] = companies_df['marketcap'].apply(parse_marketcap)
```

```python
# calcular o marketcap total e imprimir
total_marketcap_billion = companies_df['marketcap_billion'].sum()
```

```python
print(f"Valor total estimado das companhias: {total_marketcap_billion:,.2f} Bilhões de dólares")
```
