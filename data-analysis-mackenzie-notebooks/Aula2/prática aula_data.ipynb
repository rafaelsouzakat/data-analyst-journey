```python
import pandas as pd
```

```python
# Defina o caminho do arquivo no Google Drive
file_path = '/content/drive/MyDrive/Aulas_Análise_de_dados/Aula_pratica_technology/technology_companies.csv'
```

```python
# Carregar o conjunto de dados
df = pd.read_csv(file_path)
```

```python
print(df.head())
```

```python
df = df.apply(lambda x: x.str.lower() if x.dtype == "object" else x)
```

```python
print(df.head())
```

```python
df.to_csv('technology_companies_modificado.csv', index=False)
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

```python
# Converter a coluna 'MarketCap' para tipo numérico
# Remover cifrão, 't' e 'b' da coluna 'MarketCap' antes da conversão
df['MarketCap'] = df['MarketCap'].str.replace('$', '', regex=False)
df['MarketCap'] = df['MarketCap'].str.replace(' t', '', regex=False) # Remove 't' representing trillions
df['MarketCap'] = df['MarketCap'].str.replace(' b', '', regex=False) # Remove 'b' representing billions
df['MarketCap'] = df['MarketCap'].str.replace(' m', '', regex=False) # Remove 'm' representing millions

# Converter para numérico
df['MarketCap'] = pd.to_numeric(df['MarketCap'])

# Verificar o DataFrame após a conversão
print(df)
```
