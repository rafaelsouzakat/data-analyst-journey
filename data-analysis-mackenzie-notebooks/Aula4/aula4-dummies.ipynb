Calculando variáveis indicadoras/dummy

```python
import pandas as pd

# Cria um DataFrame de exemplo com uma coluna categórica
df = pd.DataFrame({
    'Animal': ['Cachorro', 'Gato', 'Pássaro', 'Cachorro', 'Gato']
})

print("DataFrame Original:")
print(df)
```

```python
# Converte a coluna 'Animal' em variáveis dummy
df_dummies = pd.get_dummies(df, columns=['Animal'])

print("\nDataFrame com Variáveis Dummy:")
print(df_dummies)
```

```python
df_dummies = pd.get_dummies(df, columns=['Animal'], prefix='Tipo')
print(df_dummies)
```

```python
df_dummies.dtypes
```

```python
df_dummies.astype(int)
```

```python

# Cria um DataFrame com filmes e gêneros
data = {
    'Movie_id': [1, 2, 3, 4],
    'title': ['Movie A', 'Movie B', 'Movie C', 'Movie D'],
    'genres': [['Action', 'Adventure'], ['Action', 'Sci-Fi'], ['Drama'], ['Sci-Fi', 'Adventure', 'Action']]
}

movies_df = pd.DataFrame(data)

print("DataFrame Original:")
print(movies_df)
```

```python
# Explode a coluna 'genres' para criar uma linha para cada gênero
movies_exploded = movies_df.explode('genres')

print("\nDataFrame Explodido:")
print(movies_exploded)

# Aplica get_dummies para criar variáveis dummy para os gêneros
movies_dummies = pd.get_dummies(movies_exploded, columns=['genres'])

# Reagrupar por filme e somar as colunas dummy
movies_dummies = movies_dummies.groupby('Movie_id').sum().reset_index()

print("\nDataFrame com Variáveis Dummy (0 e 1):")
print(movies_dummies)
```

```python
import pandas as pd

# Cria um DataFrame de exemplo com salários
data = {
    'Employee': ['John', 'Jane', 'Bob', 'Alice', 'Eve'],
    'Salary': [45000, 55000, 70000, 80000, 95000]
}

df = pd.DataFrame(data)

print("DataFrame Original:")
print(df)
```

```python
# Define os intervalos de salários e rótulos para as categorias
bins = [40000, 50000, 60000, 70000, 80000, 100000]
labels = ['40k-50k', '50k-60k', '60k-70k', '70k-80k', '80k-100k']

# Cria uma nova coluna 'Salary_Bracket' com as categorias
df['Salary_Bracket'] = pd.cut(df['Salary'], bins=bins, labels=labels)

print("\nDataFrame com Faixas Salariais:")
print(df)
```

```python
# Aplica get_dummies para criar variáveis dummy para as faixas salariais
df_dummies = pd.get_dummies(df, columns=['Salary_Bracket'])

print("\nDataFrame com Variáveis Dummy (0 e 1):")
print(df_dummies)
```

```python
import requests
import pandas as pd

# URL da API JSONPlaceholder
users_url = 'https://jsonplaceholder.typicode.com/users'
posts_url = 'https://jsonplaceholder.typicode.com/posts'

# Função para obter dados de uma URL
def fetch_data(url):
    response = requests.get(url)
    response.raise_for_status()  # Verifica se ocorreu um erro
    return response.json()

# Obter dados dos usuários
users_data = fetch_data(users_url)
print("Dados dos usuários:")
print(users_data)

# Obter dados dos posts
posts_data = fetch_data(posts_url)
print("\nDados dos posts:")
print(posts_data)

# Converter dados para DataFrame pandas
users_df = pd.DataFrame(users_data)
posts_df = pd.DataFrame(posts_data)

print("\nDataFrame dos usuários:")
print(users_df.head())

print("\nDataFrame dos posts:")
print(posts_df.head())
```

```python
import pandas as pd

# Criar um DataFrame fictício
data = {
    'CustomerID': [1, 2, 3, 4],
    'Name': ['Alice Johnson', 'Bob Smith', 'Charlie Brown', 'David Wilson'],
    'City': ['New York', 'Los Angeles', 'New York', 'Chicago'],
    'PurchaseAmount': [200, 150, 300, 250],
    'Gender': ['Female', 'Male', 'Male', 'Male'],
    'Membership': ['Premium', 'Standard', 'Standard', 'Premium']
}

df = pd.DataFrame(data)
print(df)
```

```python
# Criar variáveis dummy para 'City' e 'Membership'
df_dummies = pd.get_dummies(df, columns=['City', 'Membership'])
print("\nDataFrame com variáveis dummy:")
print(df_dummies)
```

```python
# Manipulação de strings
# 1. Separar nomes em 'FirstName' e 'LastName'
df[['FirstName', 'LastName']] = df['Name'].str.split(' ', n=1, expand=True) # Changed this line
df.drop('Name', axis=1, inplace=True)

# 2. Converter todos os nomes para minúsculas
df['FirstName'] = df['FirstName'].str.lower()
df['LastName'] = df['LastName'].str.lower()

print("\nDataFrame após manipulação de strings:")
print(df)
```

```python
# Suponha que há espaços extras na coluna 'Gender'
df['Gender'] = df['Gender'].str.strip()

# Verificar e tratar valores ausentes
df.fillna('Unknown', inplace=True)

print("\nDataFrame após limpeza adicional:")
print(df)
```
