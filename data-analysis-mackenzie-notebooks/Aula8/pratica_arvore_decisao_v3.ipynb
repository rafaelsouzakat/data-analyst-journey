# 4) Agrupar e visualizar a taxa de sobrevivência por classe.

```python
# Barras simples: taxa de sobrevivência por classe

total_sobreviventes = df_titanic.groupby('Pclass')['Survived'].sum()
# escolhi a soma porque sobrevivencia é um dado binário (Sim ou Não) que está representado como 0 ou 1

fig, ax = plt.subplots(figsize=(6, 5))

barras = ax.bar(total_sobreviventes.index, total_sobreviventes.values,
                color=['#bfbfbf', '#bfbfbf', '#c0392b'], width=0.6)

ax.bar_label(barras) # não precisa de formatação porque é int.
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(['1ª classe', '2ª classe', '3ª classe'])
ax.set_ylabel('Número de Sobreviventes')
ax.set_title('Número de Sobreviventes por Classe')
plt.show()
```

- **Como chegar na porcentagem**

A taxa é sobreviventes / total da classe. Você poderia calcular um sum() e um count() e dividir um pelo outro, e funcionaria. Mas tem um atalho que vale entender: a média de uma coluna de 0 e 1 é exatamente a proporção de 1.

Por exemplo, com 4 passageiros [1, 0, 1, 1]: a soma é 3, a contagem é 4, e a média é 3/4 = 0,75, ou seja, 75% sobreviveram. A média já faz a soma e a divisão de uma vez.

```python
# representando os sobreviventes como porcentagem do total de passageiros por classe

porcentagem_sobreviventes = df_titanic.groupby('Pclass')['Survived'].mean()
porcentagem_sobreviventes = porcentagem_sobreviventes * 100
# multiplica para converter para porcentagem

fig, ax = plt.subplots(figsize=(6, 4))

barras = ax.bar(porcentagem_sobreviventes.index, porcentagem_sobreviventes.values,
                color=['#bfbfbf', '#bfbfbf', '#c0392b'], width=0.6)

ax.bar_label(barras, fmt='%.1f%%') # formatação para porcentagem
ax.set_ylim(0, 100) # fixa o eixo y de 0 a 100, o que deixa claro que é uma porcentagem
ax.set_xticks([1, 2, 3])
ax.set_xticklabels(['1ª classe', '2ª classe', '3ª classe'])
ax.set_ylabel('Taxa de Sobrevivência (%)')
ax.set_title('Taxa de Sobrevivência por Classe')
plt.show()
```

Insights:

A taxa cai conforme a classe piora (63% → 47% → 24%), ou seja, a classe social teve forte relação com a chance de sobreviver.

```python
# agrupar os raros numa categoria 'Other' antes do get_dummies

filter_rare_titles = (df_titanic['Title'] != 'Mr') & (df_titanic['Title'] != 'Miss') & (df_titanic['Title'] != 'Mrs') & (df_titanic['Title'] != 'Master')
df_titanic.loc[filter_rare_titles, 'Title'] = 'Other'

df_titanic['Title'].value_counts()
```
