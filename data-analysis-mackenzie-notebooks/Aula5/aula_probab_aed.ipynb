```python
import numpy as np
import matplotlib.pyplot as plt

# Definir uma função para simular lançamentos de moeda
def simular_lancamentos(n):
    # Simular n lançamentos de moeda (0 para "cara", 1 para "coroa")
    resultados = np.random.randint(0, 2, size=n)

    # Calcular a frequência relativa
    frequencia_cara = np.sum(resultados == 0) / n
    frequencia_coroa = np.sum(resultados == 1) / n

    return frequencia_cara, frequencia_coroa

# Número de lançamentos
n_lancamentos = 1000

# Simular lançamentos e obter frequências
frequencia_cara, frequencia_coroa = simular_lancamentos(n_lancamentos)

# Exibir as frequências
print(f"Frequência de 'cara': {frequencia_cara:.2f}")
print(f"Frequência de 'coroa': {frequencia_coroa:.2f}")

# Criar gráfico de barras para visualização
labels = ['Cara', 'Coroa']
frequencias = [frequencia_cara, frequencia_coroa]

plt.figure(figsize=(8, 6))
plt.bar(labels, frequencias, color=['blue', 'red'])
plt.xlabel('Resultado')
plt.ylabel('Frequência Relativa')
plt.title(f'Frequência Relativa de Resultados em {n_lancamentos} Lançamentos de Moeda')
plt.ylim(0, 1)  # Ajustar o limite do eixo y para melhor visualização
plt.show()
```

```python
import numpy as np
from scipy.stats import norm

# Definir os parâmetros
xbar = 62          # Média amostral
mu0 = 60           # Média sob a hipótese nula
var = 25           # Variância
sigma = np.sqrt(var)  # Desvio padrão
n = 16           # Tamanho da amostra

# Calcular o valor de z
z = (xbar - mu0) / (sigma / np.sqrt(n))

# Calcular o valor-p
p = 2 * norm.cdf(-abs(z))

# Mostrar os resultados
print(f"Valor de z: {z:.4f}")
print(f"Valor-p: {p:.4f}")
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Carregar o dataset de diamantes
diamonds = sns.load_dataset('diamonds')
```

```python
# Mostrar as primeiras linhas do dataset
print(diamonds.head())
```

```python
# Informações gerais sobre o dataset
print(diamonds.info())

# Estatísticas descritivas
print(diamonds.describe())
```

```python
# Histograma da variável 'price'
plt.figure(figsize=(10, 6))
sns.histplot(diamonds['price'], bins=50, kde=True)
plt.title('Distribuição do Preço dos Diamantes')
plt.xlabel('Preço')
plt.ylabel('Frequência')
plt.show()
```

```python
# Contagem dos diferentes tipos de corte
plt.figure(figsize=(10, 6))
sns.countplot(x='cut', data=diamonds)
plt.title('Distribuição do Tipo de Corte')
plt.xlabel('Tipo de Corte')
plt.ylabel('Contagem')
plt.show()
```

```python
# Relação entre 'carat' e 'price'
plt.figure(figsize=(10, 6))
sns.scatterplot(x='carat', y='price', data=diamonds)
plt.title('Relação entre Carat e Preço')
plt.xlabel('Carat')
plt.ylabel('Preço')
plt.show()
```

```python
# Preço dos diamantes por tipo de corte
plt.figure(figsize=(12, 6))
sns.boxplot(x='cut', y='price', data=diamonds)
plt.title('Distribuição do Preço por Tipo de Corte')
plt.xlabel('Tipo de Corte')
plt.ylabel('Preço')
plt.show()
```

```python
# Matriz de dispersão para variáveis numéricas
sns.pairplot(diamonds[['carat', 'depth', 'price']], diag_kind='kde')
plt.show()
```

```python
# Calcular e visualizar a matriz de correlação
plt.figure(figsize=(10, 8))
corr = diamonds[['carat', 'depth', 'price', 'table']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Matriz de Correlação')
plt.show()
```
