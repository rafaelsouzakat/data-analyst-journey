- Modificando a variável Inst para ordem

```python
empresa_df.tail(10)
```

```python
# 1. Definir a ordem lógica correta das categorias
ordem_educacao = ['1o Grau', '2o Grau', 'Superior']

# 2. Converter a coluna para categorias ordenadas
empresa_df['Inst'] = pd.Categorical(empresa_df['Inst'], categories=ordem_educacao, ordered=True)
```

```python
empresa_df.dtypes
# agora a coluna 'Inst' é do tipo category!
```

```python
empresa_df.sort_values('Inst')
# agora podemos ordenar pela categoria 'Inst'!
```
