**Preencher valores ausentes de idade com a mediana da coluna é uma boa alternativa.**

Porque a mediana? Porque a mediana é robusta contra outliers, diferente da média, que pode ser facilmente enviesada por causa de outliers (pessoas muito novas ou pessoas muito velhas).

```python
titanic_df['Age'].describe()
```

```python
age_mean = titanic_df['Age'].mean()
age_median = titanic_df['Age'].median()

print('A média das idades é igual a: ', age_mean)
print('A mediana das idades é igual a: ', age_median)
```

```python
titanic_df['Age'] = titanic_df['Age'].fillna(value=age_median)

# preenchendo as idades vazias com a mediana
```

```python
titanic_df['Age'].isnull().sum()
# agora temos 0 valores ausentes para as idades!
```

```python
titanic_df.loc[titanic_df['Fare'].isna()]

# esse é o único valor ausente da coluna!
```

```python
pclass_grp = titanic_df.groupby('Pclass')
```

```python
pclass_grp.get_group(3).describe()

# Como a 'Fare' ausente é de um passageiro de terceira classe,
# agrupei as classes para calcular a mediana das 'Fares' dessa classe
```

```python
third_class_mean = pclass_grp['Fare'].get_group(3).mean()
third_class_median = pclass_grp['Fare'].get_group(3).median()

print('A média das Fares da terceira classe é: ', third_class_mean)
print('A mediana das Fares da terceira classe é: ', third_class_median)
# aqui podemos ver como os outliers afetam de forma significativa a média!
```

```python
titanic_df['Fare'] = titanic_df['Fare'].fillna(value=third_class_median)
# agora estou preenchendo o valor ausente da coluna 'Fare' com a mediana da terceira classe
```

```python
titanic_df.loc[152]
# agora a 'Fare' está preenchida!
```

```python
titanic_df.loc[titanic_df['Cabin'].isna()]
# praticamente 80% dos valores para a cabine estão vazios!
```

```python
# acredito que devo remover a coluna 'Cabin', pois preenchê-la não nos daria informação confiável.
# Portanto, vou optar por remover do DataFrame.

titanic_df = titanic_df.drop(columns=['Cabin'])
```
