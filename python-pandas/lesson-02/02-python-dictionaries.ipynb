-- Python Dictionary examples

```python
person = {
    "first": "Corey",
    "last": "Schafer",
    "email": "CoreyMSchafer@gmail.com"
}
```

```python
people = {
    "first": ["Corey"],
    "last": ["Schafer"],
    "email": ["CoreyMSchafer@gmail.com"]
}
```

```python
people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", "Doe", "Doe"],
    "email": ["CoreyMSchafer@gmail.com", "JaneDoe@gmail.com", "JohnDoe@gmail.com"]
}
```

```python
people["email"]
# dataframes are similar to this, but they have more functionality than what we have in standard python dictionaries.
```

-- Importing pandas library

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['email']
```

```python
type(df['email'])
# this means that the data type of the column is a pandas series object, which is similar to a list in python.
# a dataframe is basically a container of multiple series objects, where each series object is a column in the dataframe.
```

```python
df.email
```

```python
df[['last', 'email']]
# by doing this, we can access multiple columns in a dataframe, and the result will be a new dataframe with only the selected columns.
# it's just like the SQL SELECT statement
```

```python
df.columns
```

```python
df.iloc[0]
# iloc allows us to access rows by integer location
# here, the indexes become the column names
```

```python
df.iloc[[0, 1]]
# by doing this, we can access both the first and second rows of the dataframe, and the result will be a new dataframe with only the selected rows.
# be sure to put an inner list inside the brackets
```

```python
df.iloc[[0, 1], 2]
# if we think of loc and iloc as functions, the first argument are the rows that we want to select, and the second argument are the columns that we want to select.
```

```python
df
```

```python
df.loc[0]
# with loc we're going to be searching rows by labels
```

```python
df.loc[[0, 1], 'email']
```

```python
df.loc[[0, 1], ['email', 'last']]
```
