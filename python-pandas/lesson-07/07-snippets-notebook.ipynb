```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Adam'],
    'last': ['Schafer', 'Doe', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@email.com', 'JaneDoe@email.com', 'JohnDoe@email.com', 'A@email.com']
}

# a new name was added: Adam Doe with email A@email.com
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.sort_values(by='last')
```

```python
df.sort_values(by='last', ascending=False)
```

```python
# for example, if we wanted to sort by last name and then first name, we could do the following:
df.sort_values(by=['last', 'first'], ascending=False)
```

```python
df.sort_values(by=['last', 'first'], ascending=[False,True], inplace=True)
# this sorts the DataFrame by last name in descending order, and then by first name in ascending order

df
```

```python
df.sort_index()
```

```python
df['last'].sort_values()
# this will show us a sorted Series of the last names
```
