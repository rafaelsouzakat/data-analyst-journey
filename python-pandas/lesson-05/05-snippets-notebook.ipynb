```python
people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", "Doe", "Doe"],
    "email": ["CoreyMSchafer@gmail.com", "JaneDoe@gmail.com", "JohnDoe@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns
```

```python
df.columns = ['first_name', 'last_name', 'email']
```

```python
df
```

```python
df.columns = [x.upper() for x in df.columns]
# this is a list comprehension that loops through each column name and applies the upper() method to it
df
```

```python
df.columns = df.columns.str.replace(' ', '_')
# replace any spaces with underscores in the column names
df
```

```python
df.columns = [x.lower() for x in df.columns]
df
```

```python
df.rename(columns={'first_name': 'first', 'last_name': 'last'}, inplace=True)
# this will rename only the columns specified in the dictionary, and leave the rest of the columns unchanged
df
```

```python
df.loc[2] = ['John', 'Smith', 'JohnSmith@email.com']
df
```

```python
df.loc[2, ['last', 'email']] = ['Doe', 'JohnDoe@email.com']
df
```

```python
df.loc[2, 'last'] = 'Smith'
df
```

```python
df.at[2, 'last'] = 'Doe'
df
```

```python
# common mistakes when people try to change a single value without indexers
filt = (df['email'] == 'JohnDoe@email.com')
df[filt]['last'] = 'Smith'  # this will not work, it will give a SettingWithCopyWarning
```

```python
df
```

```python
# this is the correct way to do it
filt = (df['email'] == 'JohnDoe@email.com')
df.loc[filt, 'last'] = 'Smith'
df
```

-- More advanced way that we can update multiple rows of data

4 popular methods:

1. apply
2. map
3. applymap
4. replace

```python
# apply
# can work with either a dataframe or a series object, and it can be used to apply a function to each element in a series or each series of a dataframe
df['email'].apply(len)  # this will return a series with the length of each email address
```

```python
def update_email(email):
    return email.upper()
```

```python
df['email'].apply(update_email)  # this will return a series with the email addresses in uppercase
```

```python
df['email'] = df['email'].apply(update_email)
df
```

Lambda functions are basically anonymous functions with or without a specific name

```python
df['email'] = df['email'].apply(lambda x: x.lower())
df
```

```python
# using apply with a dataframe
df['email'].apply(len)
```

```python
df.apply(len)
# this will return the number of rows in each column
```

```python
len(df['email'])  # this will return the number of rows in the email column
```

```python
df.apply(len, axis='columns')  # this will return the number of columns in each row
```

```python
df.apply(pd.Series.min)  # this will return the minimum value (in alphabetical order) in each column
```

```python
df.apply(lambda x: x.min()) # this will give us the same result as the previous example
```

```python
# this is map, applymap doesn't work
df.map(len)
```

```python
df.map(str.lower)
```

```python
df['first'].map({'Corey': 'Chris', 'Jane': 'Mary'})  # this will return a series with the first names replaced according to the dictionary
# this will replace the name 'John' with NaN because it is not in the dictionary
```

```python
# replace method
df['first'].replace({'Corey': 'Chris', 'Jane': 'Mary'})
# using replace doesn't keeps the name 'John'
```

```python
df['first'] = df['first'].replace({'Corey': 'Chris', 'Jane': 'Mary'})
df
```
