```python
people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", "Doe", "Doe"],
    "email": ["CoreyMSchafer@gmail.com", "JaneDoe@gmail.com", "JohnDoe@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns
```

```python
df.columns = ['first_name', 'last_name', 'email']
```

```python
df
```

```python
df.columns = [x.upper() for x in df.columns]
# this is a list comprehension that loops through each column name and applies the upper() method to it
df
```

```python
df.columns = df.columns.str.replace(' ', '_')
# replace any spaces with underscores in the column names
df
```

```python
df.columns = [x.lower() for x in df.columns]
df
```

```python
df.rename(columns={'first_name': 'first', 'last_name': 'last'}, inplace=True)
# this will rename only the columns specified in the dictionary, and leave the rest of the columns unchanged
df
```

```python
df.loc[2] = ['John', 'Smith', 'JohnSmith@email.com']
df
```

```python
df.loc[2, ['last', 'email']] = ['Doe', 'JohnDoe@email.com']
df
```

```python
df.loc[2, 'last'] = 'Smith'
df
```

```python
df.at[2, 'last'] = 'Doe'
df
```

```python
# common mistakes when people try to change a single value without indexers
filt = (df['email'] == 'JohnDoe@email.com')
df[filt]['last'] = 'Smith'  # this will not work, it will give a SettingWithCopyWarning
```

```python
df
```

```python
# common mistakes when people try to change a single value without indexers
filt = (df['email'] == 'JohnDoe@email.com')
df.loc[filt, 'last'] = 'Smith'
df
```
