```python
people = {
    "first": ["Corey", 'Jane', 'John'],
    "last": ["Schafer", "Doe", "Doe"],
    "email": ["CoreyMSchafer@gmail.com", "JaneDoe@gmail.com", "JohnDoe@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
filt = (df['last'] == 'Doe')    
```

```python
df[filt]
```

```python
df.loc[filt]
```

```python
df.loc[filt, 'email']
```

```python
filt_2 = (df['last'] == 'Doe') & (df['first'] == 'John')
```

```python
df.loc[filt_2, 'email']
```

```python
filt_3 = (df['last'] == 'Schafer') | (df['first'] == 'John')
```

```python
df.loc[filt_3, 'email']
```

```python
df.loc[~filt_3, 'email']
# the simbol ~ means NOT, so this will return all the emails that are NOT in the filter filt_3
```
