-- Creating a simple dataframe with a python dictionary

```python
people = {
    "first" : ["Corey", "Jane", "John"],
    "last" : ["Schafer", "Doe", "Doe"],
    "email" : ["CoreyMSchafer@gmail.com", "JaneDoe@gmail.com", "JohnDoe@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['email']
```

```python
df.set_index('email')
# by doing this, we are turning the email values into the indexes of the DataFrame.
```

```python
df
# the dataframe is still the same, because pandas doesn't do a lot of these kind of changes inplace unless we explicitly tell it to do so
```

```python
df.set_index('email', inplace=True)
```

```python
df
```

```python
df.index
```

```python
df.loc['CoreyMSchafer@gmail.com']
# now we can search rows by their index labels instead of their row numbers!
```

```python
df.loc['CoreyMSchafer@gmail.com', 'last']
```

```python
df.iloc[0]
```

```python
df.reset_index(inplace=True)
df
# we can reset the index back to the default integer index!
```
