```python
people = {
    'first': ['Corey', 'Jane', 'John'],
    'last': ['Schafer', 'Doe', 'Doe'],
    'email': ['CoreyMSchafer@email.com', 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

- Adding columns

```python
df['first'] + ' ' + df['last']
# we're getting a Series with the first and last names concatenated with a space between them.
```

```python
# we're creating a new column and assigning it the concatenated first and last names
df['full_name'] = df['first'] + ' ' + df['last']  # creating the new column with concatenated names
df  # displaying the updated DataFrame
```

- Removing columns

```python
df.drop(columns=['first', 'last'], inplace=True) # removing the 'first' and 'last' columns from the DataFrame
df  # displaying the updated DataFrame 
```

```python
# now doing the reverse: spliting the full name into first and last names
df['full_name'].str.split(' ', expand=True)  # splitting the 'full_name' column into two separate columns
```

```python
df[['first', 'last']] = df['full_name'].str.split(' ', expand=True)
df  # displaying the updated DataFrame with the split first and last names
```

- Adding and Removing rows of Data

```python
# adding a single row of data
#df.append({'first': 'Tony'}, ignore_index=True)

df.loc[len(df)] = {'first': 'Tony'} 
# Note: the append method was removed in newer versions of pandas, so I use loc instead
df
```

```python
people2 = {
    'first': ['Tony', 'Steve'],
    'last': ['Stark', 'Rogers'],
    'email': ['IronMan@avenge.com', 'Cap@avenge.com']
}

df2 = pd.DataFrame(people2)  # creating a DataFrame from the people2 dictionary
```

```python
df2
```

```python
# adding a dataframe into another dataframe
df = pd.concat([df, df2], ignore_index=True)  # combining the two DataFrames
df  # displaying the updated DataFrame
```

```python
# removing rows
df.drop(index=5)
```

```python
# now we're removing rows based on a condition. In this case, we're removing rows where the 'last' name is 'Doe'.
filt = df['last'] == 'Doe'  # creating a filter for rows where the 'last' name is 'Doe'

df.drop(index=df[filt].index)
```
